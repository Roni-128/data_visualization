```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Create a simple dataset
data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Maths": [80, 65, 90, 55, 75, 85],
    "Science": [75, 70, 95, 60, 80, 88]
}

df = pd.DataFrame(data)

print("Student Marks:")
print(df)

# Set chart style
sns.set_theme(style="whitegrid")

# 1. Bar Chart
plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="Student", y="Maths", color="skyblue")

plt.title("Student Maths Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# 2. Histogram
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Maths", bins=5, color="orange")

plt.title("Distribution of Maths Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Number of Students")
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Maths",
    y="Science",
    color="green",
    s=100
)

plt.title("Maths vs Science Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.show()

# 4. Correlation Heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(
    df[["Maths", "Science"]].corr(),
    annot=True,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

plt.title("Marks Correlation Heatmap")
plt.show()

print("Project completed successfully!")
```
